# Split
In this section we will split the train.csv file into X_train, X_val, y_train, y_val x_val and y_val represent the test set, named this way to distinguishe between this test and test.csv file downloaded from kaggle. We will apply the ordinary split 80% of data for training and 20% for testing, and we sill preserve the balance of the two classes (survived and died) along this split.

```python
X = data.drop(columns=['Survived'])
y = data['Survived']

X_train, X_val, y_train, y_val = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print(y_train.value_counts(normalize=True))
print(y_val.value_counts(normalize=True))
```

*Notes*

A few things to notice about this code:

- **stratify= y** forces the split to preserve the same class proportions (~62% died / ~38% survived) in both X_train/y_train and X_val/y_val. Without it, a random split could — just by chance, especially with smaller val sets — land you with a val set that's, say, 70/30 or 50/50, which wouldn't accurately represent the real class balance anymore. That mismatch would make the F1 score on val misleading.
- **random_state= 42** makes the split reproducible; same rows end up in train/val every time you rerun this.
- **test_size= 0.2** 80/20 split, matching what we'd settled on earlier.
- The two print lines let us verify the stratification worked — both should show roughly 62%/38%.
